## Cell 1: imports and folder check ##

```python
import pandas as pd
import numpy as np
from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 200)

RAW = Path("../data/raw")
print("Raw data folder contents:")
for sub in sorted(RAW.iterdir()):
    print(f"  {sub.name}/")
```

## Cell 2: load Olist tables ##

```python
olist_dir = RAW / "olist"
orders = pd.read_csv(olist_dir / "olist_orders_dataset.csv")
order_items = pd.read_csv(olist_dir / "olist_order_items_dataset.csv")
reviews = pd.read_csv(olist_dir / "olist_order_reviews_dataset.csv")
customers = pd.read_csv(olist_dir / "olist_customers_dataset.csv")
products = pd.read_csv(olist_dir / "olist_products_dataset.csv")
payments = pd.read_csv(olist_dir / "olist_order_payments_dataset.csv")

for name, df in [("orders", orders), ("order_items", order_items), ("reviews", reviews),
                 ("customers", customers), ("products", products), ("payments", payments)]:
    print(f"{name:14s} shape={df.shape}")
```

## Cell 5: review score breakdown ##

```python
print("Review score distribution:")
print(reviews["review_score"].value_counts().sort_index())
print(f"\nReviews with text comments: {reviews['review_comment_message'].notna().sum():,}")
print(f"Average review score: {reviews['review_score'].mean():.2f}")
```

## Cell 6: women's clothing reviews ##

```python
womens_files = list((RAW / "womens_clothing").glob("*.csv"))
print("Files found:", [f.name for f in womens_files])
womens = pd.read_csv(womens_files[0])
print(f"\nShape: {womens.shape}")
print(f"Columns: {womens.columns.tolist()}")
womens.head(3)
```

## Cell 7: women's reviews profile ##

```python
print(f"Null review text: {womens['Review Text'].isna().sum()}")
print(f"Mean review length (chars): {womens['Review Text'].dropna().str.len().mean():.1f}")
print(f"\nRating distribution:")
print(womens["Rating"].value_counts().sort_index())
print(f"\nTop 5 departments:")
print(womens["Department Name"].value_counts().head())
```
